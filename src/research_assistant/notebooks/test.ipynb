```python
import time
from research_assistant.tools.search import search_web

t = time.perf_counter()
try:
    sources = search_web(
        "What are the concurrency and write-lock limitations of SQLite WAL mode compared to PostgreSQL?"
    )
    print(len(sources), [s.url for s in sources])
except Exception as e:
    print(type(e).__name__, e)
print(f"{time.perf_counter() - t:.1f}s")
```

```python
import time

from research_assistant.graph.build import build_graph

QUESTION = "What are the main differences between SQLite and PostgreSQL for a small web app?"

start = time.perf_counter()
result = build_graph().invoke({"question": QUESTION})
elapsed = time.perf_counter() - start

print(f"\nDone in {elapsed:.1f}s\n")

print("PLAN")
for s in result["plan"].subtasks:
    print(f"  - {s.description}")

print("\nFINDINGS")
for f in result["findings"]:
    print(f"\n[{f.status}] {f.subtask_id}")
    if f.error:
        print(f"  error: {f.error}")
    if f.answer:
        print(f"  answer: {f.answer[:400]}")
    for src in f.sources:
        print(f"  source: {src.url}")
```
